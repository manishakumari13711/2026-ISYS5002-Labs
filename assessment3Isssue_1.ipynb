```python
def upcoming_assessments(assessments):
    today = date.today()
    upcoming = []
    for a in assessments:
        due = datetime.strptime(a['due_date'], '%Y-%m-%d').date()
        days = (due - today).days
        if a['marks_obtained'] == '':
            upcoming.append((days, a))

    def get_days(item):
        return item[0]

    upcoming.sort(key=get_days)
    if not upcoming:
        print('No ungraded assessments to display.')
    for days, a in upcoming:
        if days < 0:
            status = f'Overdue by {abs(days)} day(s)'
        elif days == 0:
            status = 'Due today'
        else:
            status = f'Due in {days} day(s)'

        print(
            f"{a['unit']} - {a['assessment']}: "
            f"{status}; weight {a['weight']}%"
        )

    print(
        'Ungraded does not necessarily mean unsubmitted. '
        'Check your submission records.'
    )
```
