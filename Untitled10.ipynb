```python
def upcoming_assessments(assessments):
    today = date.today()
    upcoming = []
    for a in assessments:
        due = datetime.strptime(a['due_date'], '%Y-%m-%d').date()
        days = (due - today).days
        if a['marks_obtained'] == '' and days >= 0:
            upcoming.append((days, a))

    def get_days(item):
        return item[0]

    upcoming.sort(key=get_days)
    if not upcoming:
        print('No upcoming ungraded assessments.')
    for days, a in upcoming:
        print(f"{a['unit']} - {a['assessment']}: {days} day(s) left; weight {a['weight']}%")
```
